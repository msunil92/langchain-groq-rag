```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/essentials.txt")
docs = loader.load()
print(docs[0].page_content[:100])
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap=10, separator="\n")
chunks = text_splitter.split_documents(docs)


for chunk in chunks:
    print(f"Chunk {chunks.index(chunk)} and Chunk count: {len(chunk.page_content)}")
    print(chunk.page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=10,
    length_function=len,
    separators=["\n\n", "\n", " ", ""]
)
chunks = splitter.split_documents(docs)

chunk = chunks[0].page_content
chunk
```
